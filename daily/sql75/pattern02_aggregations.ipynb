# Beginner Level

### Q11. Count orders per customer 

```sql
-- Customer ID was available in orders itself. So no need of join

SELECT c.customer_id, count(o.order_id)
from customers c
JOIN orders o
ON c.customer_id = o.customer_id
GROUP BY c.customer_id
-- ORDER BY count(o.order_id) DESC
-- LIMIT 10
```

```sql
SELECT customer_id, COUNT(*) AS order_count
FROM orders
GROUP BY customer_id;
```

### Q12. Total sales by product

# Intermediate Level


### Q14. Summary statistics per product category

```sql
SELECT
  category,
  count(product_id) as total_products,
  SUM(inventory) as total_inventory,
  Avg(price) as avg_price,
  min(price) as min_price,
  max(price) as max_price
FROM products
GROUP BY category;
```

### Q15. Find customers with total purchases > $1000

```sql
SELECT customer_id, sum(amount) as total_amount
FROM orders
GROUP BY customer_id
HAVING total_amount > 1000
ORDER BY total_amount DESC
```

### Q16. Find average of daily order counts

```sql
SELECT avg(total_orders)
from
    (SELECT date(order_date), count(order_id) as total_orders
    FROM orders
    group by date(order_date))
```

# Advanced Level

### Q17. Sales by category and subcategory

```sql
SELECT category, subcategory, 
count( product_id) as total_products,
sum(sales) as total_sales
FROM products
GROUP BY category, subcategory
ORDER BY category, total_sales DESC
```

```sql
-- Task: Sales by category and subcategory

SELECT 
  category,
  subcategory,
  COUNT(DISTINCT product_id) AS product_count,
  SUM(sales) AS total_sales
FROM products
GROUP BY category, subcategory
ORDER BY category, total_sales DESC;
```
